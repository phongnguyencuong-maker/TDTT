nếu đúng thì tính diện tích của tam giác đó, in ra với 1 chữ số phần thập phân, nếu nó 
không phải 3 cạnh của tam giác thì in ra “Khong phai 3 canh cua tam giac”.

```python
a= float(input())
b= float(input())
c= float(input())
if a+b>c and a+c>b and b+c>a:
    print("3 cạnh tam giác")
    p=(a+b+c)/2
    s= (p*(p-a)*(p-b)*(p-c))**0.5
    print("Dien Tich =", round(s, 1))
else:
    print("Không phải 3 cạnh của tam giác")
```

Input

Output

Input

Output

```python
ten = input("Ten chu ho: ")
cs_truoc = int(input("Chi so thang truoc: "))
cs_nay = int(input("Chi so thang nay: "))
so_dien = cs_nay - cs_truoc
tien_dien = 0
if so_dien <= 50:
    tien_dien = so_dien * 1984
elif so_dien <= 100:
    tien_dien = (50 * 1984) + ((so_dien - 50) * 2050)
elif so_dien <= 200:
    tien_dien = (50 * 1984) + (50 * 2050) + ((so_dien - 100) * 2380)
elif so_dien <= 300:
    tien_dien = (50 * 1984) + (50 * 2050) + (100 * 2380) + ((so_dien - 200) * 2998)
elif so_dien <= 400:
    tien_dien = (50 * 1984) + (50 * 2050) + (100 * 2380) + (100 * 2998) + ((so_dien - 300) * 3350)
else:
    tien_dien = (50 * 1984) + (50 * 2050) + (100 * 2380) + (100 * 2998) + (100 * 3350) + ((so_dien - 400) * 3460)
tong_tien = round(tien_dien * 1.08)
print(f"Ho va ten: {ten}")
print(f"Tien phai tra la: {tong_tien}")
```

9. Chuyển đổi số thành chữ số: Nhập một số nguyên từ 0 đến 9, in ra tên chữ số 
tương ứng (0 -> "Không", 1 -> "Một", ...) – Xem trước match …case

```python
tenso= ["Không", "Một", "Hai", "Ba", "Bốn", "Năm", "Sáu", "Bảy", "Tám", "Chín"]
so = int(input("Nhập vào một số nguyên từ 0 đến 9: "))
if 0 <= so <= 9: 
    print(tenso[so])
else:
    print("Số không hợp lệ<nhập từ 0 đến 9>")
```

```python
#cách 2 dùng match...case
so = int(input("Nhập vào một số nguyên từ 0 đến 0: "))
match so:
    case 0:
        print("Không")
    case 1:
        print("Một")
    case 2:
        print("Hai")
    case 3:
        print("Ba")
    case 4:
        print("Bốn")
    case 5:
        print("Năm")
    case 6:
        print("Sáu")
    case 7:
        print("Bảy")
    case 8:
        print("Tám")
    case 9:
        print("Chín")
    case _:
        print("Số không hợp lệ<nhập từ 0 đến 9>")
```
