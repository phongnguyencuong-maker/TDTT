Bài 4: Viết chương trình in ra hình tam giác cân có độ cao h được nhập từ bàn phím 

```python
# Nhập chiều cao h từ bàn phím
h = int(input("Nhập chiều cao h: "))

for i in range(h):
    # Dòng đầu tiên (đỉnh tam giác)
    if i == 0:
        print(" " * (h - 1) + "*")
    # Dòng cuối cùng (đáy tam giác)
    elif i == h - 1:
        print("*" * (2 * h - 1))
    # Các dòng ở giữa (vỏ tam giác rỗng)
    else:
        # Khoảng trắng bên trái, ngôi sao bên trái, khoảng trắng ở giữa, ngôi sao bên phải
        print(" " * (h - 1 - i) + "*" + " " * (2 * i - 1) + "*")
```

Bài 14: Nhập vào một dãy số nguyên, dừng nhập khi người dùng nhập -1. Sau khi nhập 
xong, in số lớn nhất, số nhỏ nhất trong những số vừa nhập. KHÔNG ĐƯỢC dùng mảng

```python
max_val=None
min_val=None
while True:
    x=int(input())
    if x==-1:
        break
    if max_val is None:
        max_val=x
        min_val=x
    else:
        if x >max_val:
            max_val=x
        if x<min_val:
            min_val=x
if max_val is None:
    print("ko co so nguyen nao")
else:
    print(max_val)
    print(min_val)
```

Bài 18: Nhập vào một số nguyên dương n, kiểm tra xem n có phải là số dạng 3^k hay 
không

```python
n=int(input())
if n<1:
    is_power_of_three=False
else:
    is_power_of_three=True
    temp=n
    while temp>1:
        if temp%3!=0:
            is_power_of_three=False
            break
        temp//=3
if is_power_of_three:
    print("so co dang 3^k")
else:
    print("so ko co dang 3^k")
```

Bài 19: Dãy số fibonacci là dãy số được định nghĩa như sau: 1, 1, 2, 3, 5, 8, 13,... với số 
kế tiếp sẽ bằng tổng hai số trước đó. Nhập vào số nguyên dương A, hãy tìm số trong 
dãy số fibonacci lớn nhất nhưng không vượt quá A

```python
A=int(input())
if A<1:
    print("khong hop le")
else:
    f1=1
    f2=1
    while True:
        fn=f1+f2
        if fn>A:
            break
        f1=f2
        f2=fn
print(f2)    
```

Bài 20: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố.

```python
n=int(input())
if n<1:
    print("khong hop le")
else:
    temp=n
    cnt=0
    while temp>0:
        digit=temp%10
        if digit==2 or digit == 3 or digit == 5 or digit == 7:
            cnt+=1
        temp//=10
    print(cnt)
```
